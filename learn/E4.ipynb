## 4.1 Basic Model Build

```r
set.seed(811)
gamam_data <- data.frame(
  ClaimCost = rgamma(100, shape = 2, scale = 500), # Dependent Variable (Y)
  Age = sample(20:60, 100, replace = TRUE),
  VehicleType = factor(sample(c("Sedan", "Truck", "SUV"), 100, replace = TRUE))
)
```

```r
gamma_model <- glm(
  ClaimCost ~ Age + VehicleType,
  data = gamam_data,
  family = Gamma(link = "log") # Log link is often preferred for interpretation
  # However the canonical link is default: the power link.
  # Best practice to specify the canonical link function & choice justification
)
summary(gamma_model)
```

**Negative Binomial Distribution**

```r
# `MASS` package required (for the glm.nb function)
pacman::p_load(MASS)

set.seed(811)
n <- 150
```

```r
adbid_usd <- runif(N, min = 0.50, max = 2.50)
creative_quality <- factor(
  sample(c("Low", "Medium", "High"), n, replace = TRUE, prob = c(0.2, 0.5, 0.3))
)
device_type <- factor(sample(c("Mobile", "Desktop"), n, replace = TRUE))
log_mu <- 0.8 + (0.3 * adbid_usd) +
  (0.6 * (creative_quality == "High")) - (0.2 * (device_type == "Mobile"))

num_clisk_to_convert <- rnegbin(n, mu = exp(log_mu), theta = 4)

nb_data_marketing <- data.frame(
  num_clisk_to_convert, adbid_usd, creative_quality, device_type
)
```

```r
nb_model_marketing <- glm.nb(
  num_clisk_to_convert ~ adbid_usd + creative_quality + device_type,
  data = nb_data_marketing, link = "log"
)

print("--- Summary of Negative Binomial Model for Marketing Clicks ---")
summary(nb_model_marketing)
```
