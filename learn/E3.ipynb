## 3.1 Combining Datasets

```r
library(pacman)
pacman::p_load(tidyverse)
```

```r
policyholders <- tibble(
  "PolicyID" = c("P001", "P002", "P003", "P004", "P005"),
  "HolderName" = c(
    "Alice Aiken", "Bob Borris", "Cathy Clark", "David Doe", "Eva Einstein"
  ),
  "Region" = c("North", "South", "East", "West", "North")
)
print("Policyholders Table:")
policyholders

premiums <- tibble(
  "PolicyID" = c("P002", "P001", "P006", "P004"),
  "LastPaymentDate" = c("2025-09-01", "2025-10-10", "2025-08-20", "2025-10-05"),
  "Status" = c("Paid", "Paid", "Pending", "Paid")
)

print("Premium Table:")
premiums
```

```r
left_join(policyholders, premiums, by = "PolicyID")
```

```r
right_join(policyholders, premiums, by = "PolicyID")
```

```r
inner_join(policyholders, premiums, by = "PolicyID")
```

```r
# Outer Join
full_join(policyholders, premiums, by = "PolicyID")
```

## 3.2 Splitting the Dataset

```r
pacman::p_load(caret)

rawdata <- mtcars
```

```r
set.seed(811)
train_indices <- createDataPartition(
  y = rawdata$mpg,
  p = 0.70,
  list = FALSE
)
```

```r
train_data <- mtcars[train_indices, ]
test_data <- mtcars[-train_indices, ]
```

```r
cat(paste("Training set size:", nrow(train_data)))
cat(paste("Testing set size:", nrow(test_data)))
```

```r
variables_to_check <- c("mpg", "hp", "wt")

train_means <- sapply(train_data[variables_to_check], mean)
test_means <- sapply(test_data[variables_to_check], mean)

comparison_table <- data.frame(
  Original_Data = sapply(rawdata[variables_to_check], mean),
  Training_Set = train_means,
  Test_Set = test_means
)
print("Mean comparision of Key Numeric Variables:")
print(comparison_table)
```
